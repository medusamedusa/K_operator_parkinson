```python
####
# Addition April 3, 2026
#####

# considering the purely element-wise product, the Hadamard,
# that I developed later


K_approximation_hadamard = np.divide(
    matrix2,
    matrix1,
    out=np.zeros_like(matrix2),
    where=matrix1 != 0
)

#K_reordered_hadamard = K_approximation_hadamard[np.ix_(reordered_indices, reordered_indices)]

plotting.plot_matrix(
    #K_reordered_hadamard,
    K_approximation_hadamard,
    #labels=reordered_labels,
    labels = atlas_labels,
    colorbar=True,
    vmax=2,
    vmin=-2,
    title="Original K (Hadamard)"
)
plotting.show()

# Updated function
def create_tensor_product_submatrix_hadamard(n, i, j):
    submatrix = np.zeros((n, n))
    #submatrix[i, j] = K_reordered_hadamard[i, j]
    submatrix[i,j] = K_approximation_hadamard[i,j]
    return submatrix

# Check tensor product submatrices
n = len(atlas_labels)
tensor_product_all_submatrices_hadamard = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        tensor_product_submatrix_hadamard = create_tensor_product_submatrix_hadamard(n, i, j)
        
        print(f"Tensor Product Submatrix for ROI {atlas_labels[i]} and {atlas_labels[j]}:")
        print(tensor_product_submatrix_hadamard)
        print("\n")

        title = f"Heatmap for Submatrix {atlas_labels[i]}-{atlas_labels[j]}"
        plotting.plot_matrix(tensor_product_submatrix_hadamard, colorbar=True, vmax=1, vmin=-0.8, title=title)
        plotting.show()

        # Accumulate
        tensor_product_all_submatrices_hadamard += tensor_product_submatrix_hadamard

# Now result_matrix is just the reconstruction
result_matrix_hadamard = tensor_product_all_submatrices_hadamard

print("Result matrix shape:", result_matrix_hadamard.shape)

# Normalize
scaled_result_matrix_hadamard = result_matrix_hadamard / np.linalg.norm(result_matrix_hadamard)
scaled_K_approximation_hadamard = K_approximation_hadamard / np.linalg.norm(K_approximation_hadamard)

# Compare
tolerance = 0.0001
is_close = np.allclose(scaled_result_matrix_hadamard, scaled_K_approximation_hadamard, atol=tolerance)

if is_close:
    print("Result matrix is close to the original K_approximation.")
else:
    print("Result matrix is not close to the original K_approximation.")

# Visualize
plotting.plot_matrix(K_approximation_hadamard, labels=atlas_labels, colorbar=True, vmax=3, vmin=-3, title="Original K_approximation (Hadamard)")
plotting.plot_matrix(result_matrix_hadamard, labels=atlas_labels, colorbar=True, vmax=3, vmin=-3, title="Reconstructed Matrix")

plotting.show()
```
